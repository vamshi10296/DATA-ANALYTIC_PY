```
str = input()
k = 0
j = len(str) - 1
is_palindrome = True

while k < j:
    if (str[k] == str[j]):
        k = k + 1
        j = j - 1
    else:
        is_palindrome = False
        break

if is_palindrome:
    print(str, "is a palindrome")
else:
    print(str, "is not a palindrome")
```

```
lst = [1,2,3,'hello','c',[2,4]]

print(lst)

#append()=>Adds a single element at the end of the list.
# If a list is appended, the entire list becomes one nested element.
lst.append(3)
print(lst)

lst.append([2,3,4,5])
print(lst)


#-------------------------------------------------------------------------------

#extend() =>Adds each element of an iterable (like a list)
# to the end of the existing list individually.
lst.extend([2,3,4,5])
print(lst)

#-------------------------------------------------------------------------------


# insert()=>inserts an element at the specified index.
# Existing elements are shifted to the right.
lst.insert(2,'vamshi')
print(lst)

#-------------------------------------------------------------------------------

#pop() -> Removes and returns the element at the given index.
# If no index is provided, it removes and returns the last element.
lst.pop()
print(lst)

#-------------------------------------------------------------------------------


#remove()=> Removes the first occurrence of the specified value from the list.
# Raises ValueError if the value is not found.

lst.remove(2)
print(lst)

#-------------------------------------------------------------------------------


#index()
print(lst.index(2))


#-------------------------------------------------------------------------------




#count()
print(lst.count(3))


#-------------------------------------------------------------------------------

#sort()

lst2=[3,2,4,5]
lst2.sort()
print(lst2)  # it return the none

#-------------------------------------------------------------------------------


```
