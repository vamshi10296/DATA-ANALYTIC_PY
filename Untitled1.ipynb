```python
import numpy as np

a = np.random.randint(2, 15, 6)
b = np.random.randint(1, 19, 6)

c = np.where(b < 2, b, a)

print("c =", c)

# Sort
print("Sorted:", np.sort(c))

# Sum
print("Sum:", np.sum(c))

# Mean
print("Mean:", np.mean(c))

# Standard Deviation
print("Std:", np.std(c))

# Minimum
print("Min:", np.min(c))

# Maximum
print("Max:", np.max(c))

# Unique values
print("Unique:", np.unique(c))




```

```python

scores = []

days = ["Sun", "Mon", "Tue", "Wed", "Thu", "Fri", "Sat"]

for day in days:
    score = int(input(f"Enter score for {day}: "))
    scores.append(score)

scores = np.array(scores)

print("Scores:", scores)
print("Sum:", np.sum(scores))
print("Max:", np.max(scores))
print("Min:", np.min(scores))
print("Mean:", np.mean(scores))
print("Std:", np.std(scores))
print("Sorted:", np.sort(scores))
print("Unique:", np.unique(scores))
```
